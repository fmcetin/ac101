# The toolkit, made solid
*Where a car's cost lives before it's sold · what R² and the p-value really say · where break-even comes from. Three pieces Session 2 moved through fast — this section makes them tools you can use.*

## One car, four events — watch A = L + E

A Model Y's cost story, in illustrative round numbers:

| Event | Cash | Inventory | Liabilities | Equity (profit) | A = L + E? |
|---|---:|---:|---:|---:|---|
| 1 · Buy the materials | −$20k | +$20k | — | — | ✓ assets just swap |
| 2 · Pay wages to build the car | −$5k | +$5k | — | — | ✓ the wage becomes part of the car's cost |
| 3 · The finished car waits — even across a year-end | — | — | — | — | ✓ nothing moves; the $25k cost *waits* |
| 4 · Sell it for $42k | +$42k | −$25k | — | **+$17k** | ✓ +17 assets = +17 equity |

Profit moves **once**, in event 4: revenue $42k − **cost of goods sold $25k** = **+$17k** — the car's whole cost, released into profit at the moment of sale.

## The message
**A cost becomes an expense only when it earns its revenue.** Building a car moves money *between assets* (cash → inventory); even the assembly wages are capitalised into the car's cost rather than expensed. Selling the car creates revenue **and** releases its cost into profit as **COGS** — in the same breath. Until the sale, profit has not moved.

And that is why **deliveries** — cars *sold* — drive COGS while **production** doesn't (R² 0.84 vs 0.15 in Session 2): the regression agreed with the accounting.

## R² — the share of the misses the line removes

With **no line**, your best guess for any month's cost is the **average**. R² asks how much better the line does:

````python
from scipy import stats

demo = pd.read_csv('data/teaching/monthly_cost_demo.csv')
d_s, d_i, d_r, d_p, d_se = stats.linregress(demo['units'], demo['total_cost'])
sst = ((demo['total_cost'] - demo['total_cost'].mean())**2).sum()   # misses around the average
sse = ((demo['total_cost'] - (d_i + d_s*demo['units']))**2).sum()   # misses around the line

fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
ybar = demo['total_cost'].mean()
pred = d_i + d_s*demo['units']
a1.scatter(demo['units'], demo['total_cost'], s=55, color=BLUE, alpha=.75, edgecolors='black', zorder=5)
a1.axhline(ybar, color='0.35', lw=2)
a1.vlines(demo['units'], ybar, demo['total_cost'], color=TESLA_RED, ls=':', lw=1.8)
a1.set_title('No line: guess with the average'); a1.set_xlabel('Units'); a1.set_ylabel('Total cost (£)')
a2.scatter(demo['units'], demo['total_cost'], s=55, color=BLUE, alpha=.75, edgecolors='black', zorder=5)
xs = np.linspace(demo['units'].min(), demo['units'].max(), 40)
a2.plot(xs, d_i + d_s*xs, color='navy', lw=2)
a2.vlines(demo['units'], pred, demo['total_cost'], color=TESLA_RED, ls=':', lw=1.8)
a2.set_title('With the line: the misses shrink'); a2.set_xlabel('Units')
for a_ in (a1, a2): a_.grid(True, alpha=.3)
plt.tight_layout(); plt.show()
````

````python
print(f"Squared misses around the average : {sst:,.0f}")
print(f"Squared misses around the line    : {sse:,.0f}")
print(f"Share the line removes            : 1 − {sse/sst:.3f} = {1 - sse/sst:.3f}")
````

**That is R².** R² = 0.841 means the line removes **84%** of the variation the average leaves behind. Formally: $R^2 = 1 - \sum(Y-\hat{Y})^2 / \sum(Y-\bar{Y})^2$. R² tells you *how much to lean on the line* — never that the line is *right* (recall B6: total revenue fitted COGS at R² = 0.963 and was still the wrong driver).

## The p-value — the luck question

Session 2's slope said **$31,951 per car**, with p = 0.0013. The p-value answers one worry: *with only 8 quarters, couldn't a slope that size appear by pure luck — even if deliveries had nothing to do with COGS?* Break the link on purpose and see:

````python
tq = pd.read_csv('data/raw/tesla_quarterly.csv')
tq = tq[tq['year'].isin([2023, 2024])].copy()
dels = tq['deliveries'].values.astype(float)
cogs = tq['cogs_millions'].values.astype(float)
ts, ti, tr, tp_, tse = stats.linregress(dels, cogs)

rng = np.random.default_rng(101)                      # seeded — rerun and you get the same picture
shuffle_slopes = np.array([stats.linregress(dels, rng.permutation(cogs)).slope
                           for _ in range(2_000)]) * 1e6
n_reached = int((np.abs(shuffle_slopes) >= abs(ts*1e6)).sum())

fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(shuffle_slopes, bins=45, color=BLUE, alpha=.75, edgecolor='white')
ax.axvline(ts*1e6, color=TESLA_RED, lw=2.5)
ax.annotate(f'our slope: ${ts*1e6:,.0f}', xy=(ts*1e6, 45), xytext=(ts*1e6 - 24_000, 72),
            fontsize=10, color=TESLA_RED, fontweight='bold',
            arrowprops=dict(arrowstyle='->', color=TESLA_RED))
ax.set_xlabel('Slope luck produced ($ per car)'); ax.set_ylabel('Shuffles')
ax.set_title('What "statistically significant" looks like')
ax.grid(True, alpha=.3)
plt.tight_layout(); plt.show()
print(f"Shuffles (out of 2,000) that reached the real slope: {n_reached}")
````

In 2,000 shuffles, luck **never once** reached our slope. The regression's **p = 0.0013** asks exactly this question with a formula instead of shuffling: *if the true slope were zero, chance would produce one this strong about once in 750 tries.* That is all **statistically significant** means — *too strong to be luck* (convention: p < 0.05). It is **not** the accuracy of the estimate, and **not** the probability the answer is right.

## Define it well
**Statistically significant** — too strong to be luck (small p). Says nothing about **size**.
**Economically significant** — big enough to change a decision. Says nothing about **certainty**.

Two illustrations: a paint-shop electricity slope of **$9 per car with p = 0.001** is certainly real and utterly irrelevant next to $31,951 of variable cost (*statistically yes, economically no*). A supplier-rebate effect of **−$2,100 per car with p = 0.31** would matter enormously if real, but eight quarters can't tell (*economically yes, statistically not yet — get more data before acting*). **Act on numbers that are both.**

## Use the output — classify cost accounts by their p-values

Regression output is *for* three things: the **split** feeds decisions (CVP, forecasts, the case); **R²** says how far to trust the line; and the **p-values classify cost behaviour**, account by account — this is how a finance team actually splits a ledger:

````python
# SIMULATED quarterly cost accounts (labelled): built on the REAL 2023–24
# deliveries, plausible magnitudes, seed fixed — Tesla does not publish this detail.
rng2 = np.random.default_rng(19)
acct = pd.DataFrame({
    'Factory security & insurance': 44 + rng2.normal(0, 2.2, 8),
    'Battery cells & materials':    0.0209 * dels + rng2.normal(0, 55, 8),
    'Plant power & maintenance':    220 + 0.00025 * dels + rng2.normal(0, 5, 8),
})
def fit_account(y):
    res = stats.linregress(dels, y)
    t_int = res.intercept / res.intercept_stderr
    p_int = 2 * (1 - stats.t.cdf(abs(t_int), len(dels) - 2))
    return res.intercept, p_int, res.slope*1e6, res.pvalue, res.rvalue**2

print(f"{'ACCOUNT':<34}{'Intercept (p)':>18}{'Slope/car (p)':>20}{'R²':>8}")
print("-" * 80)
for name in acct:
    a0, pa, b1, pb, r2 = fit_account(acct[name])
    pa_s = f"{pa:.3f}" if pa >= 0.001 else "<0.001"
    pb_s = f"{pb:.3f}" if pb >= 0.001 else "<0.001"
    print(f"{name:<34}{f'${a0:,.0f}m ({pa_s})':>18}{f'${b1:,.0f} ({pb_s})':>20}{r2:>8.2f}")
````

## Break-even, derived — not just drawn

Start from what profit *is*, using Session 2's cost function:

$$\pi \;=\; \underbrace{P \cdot Q}_{\text{revenue}} - \underbrace{(FC + VC \cdot Q)}_{\text{total cost}} \;=\; (P - VC)\cdot Q - FC \;=\; \text{CM}\cdot Q - FC$$

Set $\pi = 0$ for **break-even**: $Q^{*} = FC/\text{CM}$. Want a **target profit** $\pi^{*}$? $Q = (FC + \pi^{*})/\text{CM}$ — the pile just grows by the target. Sanity check on Tesla: $4,295m ÷ $22,648 ≈ 190,000 cars a quarter — Session 2's number, now proved rather than pictured.

````python
rev24 = tq.loc[tq['year'] == 2024, 'revenue_millions'].sum()
del24 = tq.loc[tq['year'] == 2024, 'deliveries'].sum()
rev_per_car = rev24 * 1e6 / del24
cm_unit  = rev_per_car - ts*1e6
fixed_q  = ti
be_units = fixed_q*1e6 / cm_unit
actual_q = del24 / 4

qv = np.linspace(0, actual_q*1.25, 200)
pi_m = (cm_unit*qv - fixed_q*1e6) / 1e6
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(qv/1000, pi_m, color='navy', lw=2.5)
ax.axhline(0, color='0.4', lw=1.2)
ax.fill_between(qv/1000, pi_m, 0, where=(pi_m < 0), color=TESLA_RED, alpha=.12)
ax.fill_between(qv/1000, pi_m, 0, where=(pi_m >= 0), color='seagreen', alpha=.12)
ax.scatter([0], [-fixed_q], s=55, color=TESLA_RED, zorder=5)
ax.axvline(be_units/1000, color='black', ls=':', alpha=.6)
ax.annotate(f'π = 0 at Q* ≈ {be_units/1000:.0f}k', xy=(be_units/1000, 0),
            xytext=(be_units/1000 + 18, -2600), fontsize=10)
ax.axvline(actual_q/1000, color='navy', ls='--', alpha=.6)
ax.annotate(f'actual ≈ {actual_q/1000:.0f}k', xy=(actual_q/1000, 0),
            xytext=(actual_q/1000 + 8, 1200), fontsize=10, color='navy')
ax.set_xlabel('Deliveries per quarter (000s)'); ax.set_ylabel('Quarterly profit, $m (COGS basis)')
ax.grid(True, alpha=.3)
plt.tight_layout(); plt.show()
````

# One pot of overhead, two ways to treat it
**Why overhead is the story now.** A century ago, making things was mostly *direct* cost — materials you could weigh, labour you could clock; overhead was a rounding error. Automation inverted that: robots, giga-presses, paint shops, test labs and battery plants are **indirect** — enormous, and traceable to no single car. At our stylised Fremont the shared pot comes to **\$9.2 bn a year — ≈ \$7,480 per car**, rivalling the car's own materials. (Recall **De Ridder (2024)** from Session 1: intangible, high-fixed-cost production is this pattern, economy-wide.) The biggest slice of a modern product's cost is the slice nobody can point at — and what to do with it is this session.

## The fork — define them well
- **Variable (marginal) costing:** fixed overhead is a **period expense** — all of it hits this period's profit, and no unit carries any of it.
- **Absorption (full) costing:** fixed manufacturing overhead is *baked into each unit's cost*. A unit made but **not sold** carries its slice of fixed overhead into **inventory** (the balance sheet) — not this period's expense.

**One of these is not a choice**: external financial reports **must** use absorption — IAS 2 (and US GAAP) require fixed production overhead in inventory cost. Variable costing is the internal, decision-making lens — yesterday's whole CVP toolkit was variable-costing thinking.

| | **Absorption (full)** | **Variable (marginal)** |
|---|---|---|
| Fixed production overhead | in each unit's cost | expensed in the period |
| Required for | **external reports** (IAS 2) | — (internal only) |
| Best at | long-run product economics | short-run decisions and control (CVP) |
| Weakness | profit moves with **production**, not just sales | understates the long-run cost of complexity |

Absorption's weakness deserves its own discovery:

## Discussion — the quarter is slipping
You run the Fremont plant, and reported profit uses **absorption costing**. It's quarter-end. Sales are **fixed by demand**: 120,000 cars will sell — you can't conjure buyers. On the current plan (build 120,000), reported profit comes in at **\$600m**; your **bonus gate is \$800m**; and the line has **spare capacity**.

**What could you do — entirely legally — to close a \$200m gap without selling one extra car?**

*Hypothetical scenario — realistic single-plant quarterly magnitudes: price \$45,000, variable cost \$30,000 per car, fixed production overhead \$1,200m.*

The lever someone always finds: **build more than you sell.** First the **picture** — where does fixed overhead *go* under absorption costing? Say the plant builds **150,000** cars but sells only **120,000**:

````python
# the worked example's givens — one plant, one quarter
price, var_cost, fixed_oh, sales = 45_000, 30_000, 1_200e6, 120_000
````

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

*Recall Session 1's **cost vs expense**: the parked \$240m is still a **cost** — an asset sitting in inventory — and becomes an **expense** only when those cars sell. Nothing vanished; it's just not in **this quarter's** profit. (The figure tracks the fixed overhead only — the cars' variable costs flow through COGS as usual.)*

Now watch reported profit as we **overproduce** — sales held fixed at 120,000 cars:

````python
# price, var_cost, fixed_oh, sales defined above (one plant, one quarter)
def profit(production, costing):
    revenue = price * sales
    if costing == 'variable':
        return (revenue - var_cost*sales - fixed_oh) / 1e6            # fixed all expensed
    foh_per_unit = fixed_oh / production                              # absorption: spread over output
    cogs = (var_cost + foh_per_unit) * sales                         # only SOLD units expensed
    return (revenue - cogs) / 1e6

print(f"{'Production':>12}{'Absorption profit':>20}{'Variable profit':>18}{'Parked in inventory':>22}")
print("-" * 72)
for p in (120_000, 135_000, 150_000, 180_000):
    parked = (p - sales) * (fixed_oh/p) / 1e6
    print(f"{p:>12,}{profit(p,'absorption'):>18,.0f}m{profit(p,'variable'):>16,.0f}m{parked:>20,.0f}m")
````

**135,000 isn't enough (\$733m). 150,000 clears the gate: \$840m reported** — a \$200m gap closed by building 30,000 cars nobody ordered. The variable column never moves: same sales, same economics.

````python
prods = np.arange(120_000, 182_500, 2_500)
absn = [profit(p, 'absorption') for p in prods]
varn = [profit(p, 'variable')   for p in prods]
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(prods/1000, absn, color=TESLA_RED, lw=2.5, label='Reported profit (absorption)')
ax.plot(prods/1000, varn, color=GREEN, lw=2.5, ls='--', label='True economics (variable)')
ax.axvline(sales/1000, color='gray', ls=':', label='Sales (fixed)')
ax.set_xlabel('Cars produced (000s)'); ax.set_ylabel('Reported quarterly profit ($m)')
ax.set_title('Same sales, same costs — just build more, and "profit" rises')
ax.legend(); ax.grid(True, alpha=.3)
plt.tight_layout(); plt.show()
````

## What just happened — name it
Make more than you sell, and each unsold car carries a slice of fixed overhead **off the income statement and into inventory**. Reported profit rises **without a single extra sale** — the manager hit the bonus by **building cars no one bought**. This is **earnings management through overproduction** (a *real-activities* manipulation): legal, GAAP-compliant, and value-destroying — cash is now tied up in metal, and the inventory must eventually be sold or written down.

**The floor takeaway (everyone leaves with this):** producing more than you sell can *flatter* reported profit — which is the whole reason managers also watch **variable costing**, where the trick doesn't work.

If you read Session 1's T-account aside, here is the overproduction quarter in that machinery — tracking the **fixed overhead only**; the cars' variable costs flow through COGS as usual (build 150,000, sell 120,000; fixed overhead \$1,200m at \$8,000/car):

(And if the cars never sell, the write-down lands even harder.)

**The deferral comes home.** The parked cost *waits*. A small firm, two quarters, same trick — produce 10,000 then 12,000 units; sell 8,000 then 14,000; fixed overhead £20,000 a quarter (price £5; variable manufacturing cost £1/unit; variable marketing £0.20 per unit sold; fixed marketing £5,000; FIFO):

| Reported profit | Q1 — produce > sell | Q2 — sell > produce | Both quarters |
|---|---:|---:|---:|
| **Variable costing** | £5,400 | £28,200 | £33,600 |
| **Absorption costing** | £9,400 | £24,200 | **£33,600** |
| The gap | **+£4,000** | **−£4,000** | **0** |

Q1's flattery is exactly Q2's hangover: the £4,000 parked in inventory (2,000 unsold units × £2 of fixed overhead at Q1's £20,000 ÷ 10,000 rate) lands in profit the moment those units sell. **Over the cycle the two systems agree — the trick only moves profit *between* periods**, unless production keeps climbing, which is how inventory bloat and the eventual write-down start.

## From the research (LSE tradition)
- **Burgstahler & Dichev (1997):** across thousands of firms, there are **suspiciously few small losses and too many small profits** — a tell-tale "kink" at zero, as firms nudge results just over the line.
- **Roychowdhury (2006):** firms facing a miss **really do overproduce** to push fixed cost into inventory and report higher margins — exactly the lever you just pulled.

````python
rng = np.random.default_rng(7)
e = rng.normal(0.01, 0.03, 60_000)                       # scaled earnings
mask = (e > -0.012) & (e < 0) & (rng.random(len(e)) < 0.65)   # firms nudge small losses over the line
e[mask] = rng.uniform(0.0, 0.005, size=mask.sum())            # …into small PROFITS (the tell-tale kink)
fig, ax = plt.subplots(figsize=(9, 4.2))
ax.hist(e, bins=np.arange(-0.10, 0.12, 0.005), color=BLUE, edgecolor='white')
ax.axvline(0, color=TESLA_RED, lw=2)
ax.set_xlabel('Reported earnings (scaled)'); ax.set_ylabel('Number of firms')
ax.set_title('Too few small losses, too many small profits (illustrative)')
ax.set_xlim(-0.10, 0.11); plt.tight_layout(); plt.show()
````

## Discussion

## The simplest rule: split by volume (traditional allocation)

The most common method spreads overhead by **production volume** — one rate per car (accountants call it a **plant-wide rate**), on the logic "more cars, more cost."

````python
total_oh = pools['annual_cost_millions'].sum() * 1e6        # $
total_vol = prod['annual_volume_fremont'].sum()
rate = total_oh / total_vol                                 # $ per car, one blended rate

prod['trad_oh']     = rate
prod['trad_cost']   = prod['direct_materials_per_unit'] + prod['trad_oh']
prod['trad_profit'] = prod['avg_price_usd'] - prod['trad_cost']
prod['trad_margin'] = prod['trad_profit'] / prod['avg_price_usd'] * 100

print(f"One blended overhead rate = ${rate:,.0f} per car\n")
print(f"{'Model':<12}{'Price':>9}{'Materials':>11}{'Overhead':>10}{'Profit':>10}{'Margin':>8}")
print("-" * 60)
for _, r in prod.iterrows():
    print(f"{r['Model']:<12}{'$'+format(r['avg_price_usd'], ',.0f'):>9}"
          f"{'$'+format(r['direct_materials_per_unit'], ',.0f'):>11}"
          f"{'$'+format(r['trad_oh'], ',.0f'):>10}"
          f"{'$'+format(r['trad_profit'], ',.0f'):>10}{r['trad_margin']:>7.0f}%")
````

Every model — Model Y or Model X — gets the same overhead charge.

That mismatch has a name:

## Define it well
**Cross-subsidisation:** when one product is charged *more* overhead than it really causes so that another is charged *less*. Under a single volume rate, **high-volume simple products subsidise low-volume complex ones** — the simple cars look worse than they are, the complex cars better.

## Activity-based costing (ABC): charge each cost to what drives it

ABC splits overhead into **cost pools**, each with its own **cost driver** — the thing that actually causes that cost — then charges each model for the drivers it uses.

So the rate is \$2,400 m ÷ 14.56 m hours ≈ **\$165 per assembly hour** — and a Cybertruck, at 25 hours, picks up 25 × \$165 ≈ **\$4,100** of assembly overhead alone. That's the whole ABC mechanic.

````python
# Total activity for each driver
prod['tot_assembly'] = prod['annual_volume_fremont'] * prod['assembly_hours_per_unit']
prod['tot_kwh']      = prod['annual_volume_fremont'] * prod['kwh_capacity']
prod['tot_tests']    = prod['annual_volume_fremont'] * prod['test_cycles_per_unit']
driver_tot = {'assembly_hours': prod['tot_assembly'].sum(),
              'kwh_capacity':   prod['tot_kwh'].sum(),
              'test_cycles':    prod['tot_tests'].sum(),
              'shipments':      prod['annual_volume_fremont'].sum(),
              'facility_floor': prod['annual_volume_fremont'].sum()}

# Rate for each pool = pool cost / total driver volume
rate_of = {p['cost_pool']: p['annual_cost_millions']*1e6 / driver_tot[p['cost_driver']]
           for _, p in pools.iterrows()}
print("ACTIVITY RATES (cost per unit of driver)")
for _, p in pools.iterrows():
    print(f"  {p['cost_pool']:<16} ${rate_of[p['cost_pool']]:>10,.2f} per {p['driver_unit']}")

# Allocate to each model
prod['abc_oh'] = (prod['assembly_hours_per_unit'] * rate_of['assembly']
                  + prod['kwh_capacity']          * rate_of['battery_systems']
                  + prod['test_cycles_per_unit']  * rate_of['quality_testing']
                  + rate_of['logistics'] + rate_of['facilities'])
prod['abc_profit'] = prod['avg_price_usd'] - prod['direct_materials_per_unit'] - prod['abc_oh']
prod['abc_margin'] = prod['abc_profit'] / prod['avg_price_usd'] * 100
````

Honesty caveat on the rates: logistics & facilities are charged per vehicle — our simulated logistics ships cars individually, so shipments ≈ vehicles here; in reality cars move in multi-car transporter batches, a true batch-level driver. Facilities is a facility-level cost spread by volume as a floor-space proxy, not a true square-foot driver.

## Why those drivers? The cost hierarchy

## Define it well (Cooper & Kaplan, 1991)
Overhead sits at different **levels**, and the right driver depends on the level:

| Level | Caused by… | Tesla example | Driver |
|---|---|---|---|
| **Unit** | each unit made | assembly labour, battery-plant machine time | hours, kWh |
| **Batch** | each *batch/run* | a production set-up, a delivery shipment | set-ups, shipments |
| **Product** | each *product line* existing | Cybertruck's unique tooling & engineering | per model |
| **Facility** | the *plant* existing | the Fremont building | (can't be caused away — spread by volume) |

Volume allocation charges *everything* as if it were unit-level. That's exactly why low-volume complex models (their **batch**- and **product**-level costs spread over few units) are under-costed — and why ABC, which respects the hierarchy, corrects it.

## Key takeaways
1. **How you allocate shared overhead decides which product looks profitable.** A single volume rate causes **cross-subsidisation**.
2. **ABC** charges each cost pool by the **driver** that causes it; respecting the **cost hierarchy** (unit/batch/product/facility) is *why* complex low-volume products are under-costed by volume allocation.
3. ABC changes the *distribution* of profit, not the **total** — but the distribution is the decision-relevant part. **Model Y + Model 3 are ~86% of the profit in our simulated Fremont world** — directionally right for the real Tesla, where the 3/Y are ~95% of volume.

5. **Absorption costing is mandatory for external reports** (IAS 2) — and because it parks fixed overhead in inventory, **overproduction can inflate reported profit**: legal, real, value-destroying, and self-reversing when the units finally sell. Managers watch **variable costing** alongside it.

**▸ C5 — the absorption trick.** Sales fixed at 120,000; the bonus rides on **absorption-costing** profit (fixed overhead baked into each car's cost).
Produce 120,000: fixed overhead per car = \$1,200m ÷ 120,000 = \$10,000 → (45,000 − 30,000 − 10,000) × 120,000 = **\$600m**. **Variable costing** — fixed overhead expensed in full each period, never parked in a unit — agrees: **\$600m**.
Produce 150,000, sell 120,000: the pot spreads thinner, \$1,200m ÷ 150,000 = **\$8,000** a car → (45,000 − 30,000 − 8,000) × 120,000 = **\$840m**. Variable costing still says **\$600m**.
The missing **\$240m** = 30,000 unsold cars × \$8,000, parked in **inventory** — "the cost *waits*", weaponised.
Name it: **earnings management through overproduction** (Roychowdhury 2006, *JAE*) — a *real-activities* manipulation: legal (absorption is what external reporting requires), value-destroying (cash tied up in unsold metal, write-down risk), and self-reversing when the cars finally sell.
*Slips:* computing the produce-150,000 profit on 150,000 *sold* (demand is fixed at 120,000); claiming variable-costing profit also rises; saying the \$240m "disappeared" — it sits in inventory and hits COGS when the cars sell, or worse, as a write-down.
*The line everyone leaves with:* producing more than you sell can **flatter** reported profit — which is why managers also watch variable costing, where the trick does not work.

- **Paint-shop retooling, completed last year (\$600m)** — **sunk**: a past cost never enters this period's production overhead.
- **Battery-plant electricity (\$300m)** — a **double-count**: metered separately, but already inside the \$3,200m battery pool.

12,000 − 1,900 − 600 − 300 = **\$9,200m**, and the \$7,480 rate is unchanged. **An allocation can be wrong before you pick a single driver — audit the numerator first.**
*Slips:* striking **Facilities** on "the building isn't caused by any car" — facility-level cost **is** production overhead, the hierarchy's bottom row; keeping the \$1,900m because "advertising helps sell cars"; adding the \$300m to the battery pool instead of recognising it is already inside.

## Accounting: the information system of a business

**Accounting** is how an organisation **measures, records, summarises and communicates** its economic activity. Without it, managers fly blind — *you can't manage what you can't measure.* And it is **costly** — systems, people, time. It survives for one reason: the decisions it improves are worth more than it costs. Much of it, in fact, exists for life *inside* the firm — costing, budgets, standard costs as internal "prices" — rather than for outside investors (Ball 2026, *Journal of Accounting and Economics*).

### The foundation: A = L + E

Before management accounting, one equation underlies *all* accounting. Everything a business **controls** (its assets) was funded either by **borrowing** (liabilities) or by its **owners** (equity):

$$\text{Assets} = \text{Liabilities} + \text{Equity}$$

It always balances — every pound of resource came from *somewhere*. Watch it stay balanced as you start a small business:

## Predict (show of hands)
Step 3 below spends **£2,000 of the business's cash on an espresso machine**. Before we run the cell — show of hands: **does the equation still balance? And which side moves?**

````python
steps = pd.DataFrame([
    ["Put in £5,000 of your own cash",        5000,    0, 5000, "cash 5,000"],
    ["...and borrow £3,000 from the bank",    8000, 3000, 5000, "cash 8,000"],
    ["Buy a £2,000 espresso machine",         8000, 3000, 5000, "cash 6,000 + machine 2,000"],
], columns=["Event", "Assets (£)", "Liabilities (£)", "Equity (£)", "What the assets are"])
steps["Balances? A = L + E"] = steps["Assets (£)"] == steps["Liabilities (£)"] + steps["Equity (£)"]
print(steps.to_string(index=False))
````

Buying the machine just *swaps* one asset (cash) for another (equipment) — total assets, and the equation, don't move.

<details><summary><b>▸ The machinery underneath: T-accounts (a first look)</b> — attempt first, then open</summary>

How do accountants *record* those three events so the equation can never break? Every account gets a **T**: one side for increases, the other for decreases. By centuries-old convention, **assets grow on the left** (a "debit", Dr) and **liabilities and equity grow on the right** (a "credit", Cr) — and every transaction posts **equal amounts left and right**, always touching at least two accounts. That is **double-entry bookkeeping**, and it is *why* A = L + E always balances: every entry adds the same amount to both sides of the network, so the equation can't break.

Our espresso start-up again — events **(1)** £5,000 own cash, **(2)** £3,000 loan, **(3)** £2,000 machine:

```
       CASH (asset)          EQUIPMENT (asset)      BANK LOAN (liability)     OWNER'S EQUITY
      Dr    │    Cr            Dr    │    Cr           Dr    │    Cr           Dr    │    Cr
  (1) 5,000 │              (3) 2,000 │                       │ (2) 3,000             │ (1) 5,000
  (2) 3,000 │ (3) 2,000              │                       │                       │
  ──────────┼──────────    ──────────┼──────────      ───────┼──────────      ───────┼──────────
  bal 6,000 │              bal 2,000 │                       │ bal 3,000             │ bal 5,000
```

Check: assets £6,000 + £2,000 = £8,000; liabilities £3,000 + equity £5,000 = £8,000 — the balance is *built in*, transaction by transaction.

*That's as deep as this course goes — we **use** the numbers rather than book them. But when Session 3 shows a cost "moving into inventory", this is the machinery doing it.*

</details>

## One term to get straight: cost vs expense
A **cost** is a resource you give up to get something. It becomes an **expense** when it's *used up* to earn revenue — sometimes straight away (today's advertising), sometimes later: a machine turns into expense gradually as **depreciation** (spreading the machine's cost over the years it helps earn revenue); a car's build cost waits on the shelf as **inventory** and only becomes an expense — "**cost of goods sold**" — when the car is sold. So the same £ can be a **cost (asset) now and an expense later** — keep the two ideas distinct.

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

*A cost becomes an expense when it's used up — instantly (advertising), gradually (depreciation), or **only when sold** (inventory → cost of goods sold). Session 3 leans hard on that third lane.*

The same firm keeps **three sets of books** — three *views* of the same reality for three audiences — financial (for outsiders), management (for insiders), and tax (for the authorities). Tesla's shareholders read *"FY2024 revenue $97.7bn, cost of revenues $80.2bn, operating income $7.1bn"*; Tesla's plant manager reads *"shave 10% off battery-cell cost and we save ~$500 a car."* Both true; different jobs.

## Two very different kinds of accounting

| Aspect | **Financial accounting** | **Management accounting** |
|---|---|---|
| Audience | External (investors, lenders, regulators) | Internal (managers, you) |
| Purpose | Stewardship & compliance | Decisions & control |
| Rules | Standardised (GAAP, IFRS) | Flexible — whatever helps the decision |
| Time focus | Backward (what happened) | Forward (what should we do) |
| Frequency | Quarterly, annually | Whenever a decision needs it |

## Discussion
Tesla wants to **borrow $5bn** from a bank; separately, a Tesla manager must **decide the Model Y's price**. Which kind of accounting does each need — and why?

And here is why the internal kind matters so much — **Herbert Simon's thought experiment**: a Martian points a telescope at Earth, with organisations showing up as green areas and market transactions as thin red lines. The Martian sees mostly *green* — it would never call this a "market economy". Inside the green there are no market prices to guide decisions; what firms use instead is management accounting.

Financial accounting is *required by law*. Management accounting is *chosen* by managers because it helps them run the business. It serves three jobs — **planning** ("what should we do?"), **decisions** ("which option is better?"), and **control** ("did we hit the plan?") — and they map onto your week:

# Reading Tesla's costs
## "Cost of what?" — the cost object

A **cost object** is anything you want the cost *of*: a product (Model Y), a department (the Fremont factory), a customer (a fleet buyer), a project (Cybertruck development). The same factory cost is "direct" to the factory but "indirect" to a single car — two terms we'll define properly in a few minutes — so you must fix the cost object *first*.

## One idea to carry all week
The "cost" of something **depends on the question**. The cost of a Model Y for *financial reporting* (an average) differs from its cost for *pricing* (incremental) and for *make-or-buy* (avoidable). Hold onto this — it returns every session.

We classify costs two ways, and **each answers a different manager's question.**

## Classification 1 — by behaviour: *how does profit move with volume?*

- **Variable costs** change *in total* as volume changes (battery cells: more cars, more cells).
- **Fixed costs** stay the same *in total* across a sensible range of volume — accountants call this the **relevant range** (factory insurance: the same premium at 400k or 500k cars).

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

## ▸ Outside the relevant range
The "fixed" label is only guaranteed *inside* the relevant range. Push volume far enough beyond it and fixed costs **step up** (a second assembly line, new premises) and variable costs **bend** (overtime, bottlenecks); fall far below it and each car **costs far more to build** — an underused plant, a crew still finding its rhythm on the line. The left panel's fixed-cost steps land in total cost as **jumps at both edges of the range**, and the straight line's **intercept** is *not* the true cost at zero volume — the low band runs on **$2.4bn** of fixed cost, not the extrapolated $3.0bn. Every fixed/variable call carries an implicit volume range — check it before you rely on it.

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

### The twist that trips everyone: per-unit behaviour

Flip to *per car* and the pattern inverts — and this is a classic beginner trap:

| | Total | **Per unit** |
|---|---|---|
| Variable cost | rises with volume | **constant** |
| Fixed cost | constant | **falls with volume** |

````python
fixed = 3_000_000_000           # $3bn fixed — same illustrative number as the plot above, not Tesla's actual
for n in (300_000, 500_000):
    print(f"  At {n:,} cars: fixed cost per car = ${fixed/n:,.0f}")
````

Same $3bn — but "per car" it slid from **$10,000 to $6,000**. "Fixed cost per unit" is an **artificial** number: it depends on a volume you pick.

## Your turn (then check)
You're the analyst. At 300,000 cars the accountants report a **full unit cost of $30,000 per car** ($20,000 variable + the $10,000 fixed-per-car figure we just computed). The CFO asks for the **total cost budget at 500,000 cars**. In pairs: **compute it and commit to a number** before we run the next cell.

````python
budget_wrong = (fixed/300_000 + 20_000) * 500_000 / 1e9
budget_right = (20_000 * 500_000 + fixed) / 1e9
print(f"  Budget 500k cars using the 300k unit cost: ${budget_wrong:,.1f}bn  ← WRONG")
print(f"  Correct (variable per car + total fixed):   ${budget_right:,.1f}bn  ← think in TOTALS")
````

Where did the $30,000 come from? It was a *correct* total-cost build at 300,000 cars — $3.0bn fixed + $20,000 × 300,000 = $9.0bn; ÷ 300,000 = $30,000 a car. Correct arithmetic, frozen at one volume — the error is only in *scaling* it.

**The trap:** never budget by scaling a **full** unit cost (fixed cost baked in) to a new volume. A purely *variable* unit cost is fine to scale.

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

## Misconception check
"Fixed cost per unit" feels like a real per-car number you can add to variable cost and scale up. It isn't — total fixed cost is flat; the per-unit figure is just that flat number divided by whatever volume you chose. **For decisions, think in total costs, not unit costs.** And two names worth keeping: a per-unit figure built by division is an **average cost**; the extra cost of one more car is its **marginal cost**. Session 2 leans on the first; Session 4 runs on the second.

## Classification 2 — by traceability: *what does this product cost?*

- **Direct costs** can be traced to the cost object economically (the battery pack in *this* Model Y).
- **Indirect costs (overhead)** can't be — they're shared and must be **allocated** (the factory manager's salary covers Y, 3, S and X). *How* to allocate is the whole of Session 3.

*(this figure lives on the session page — [fmcetin.com/ac101](https://fmcetin.com/ac101/))*

<details><summary><b>▸ Why these distinctions run the modern economy</b> — attempt first, then open</summary>

Software is the extreme cost structure: **huge fixed cost to build, near-zero variable cost to serve one more user**. De Ridder (2024, "Market Power and Innovation in the Intangible Economy", *American Economic Review* 114(1), 199–251 — LSE Economics) shows that as intangibles spread, this cost structure alone can explain rising markups, fewer new challengers entering, and slower productivity growth. And because ever more cost is **indirect** — shared software, data, R&D — "what does *this* product cost?" keeps getting harder to answer. That question is Session 3.
